```python
SNum = int(input("Input a number - digital for pattern generation e.g 5"))


# forward Print mechanism 
for i in range(1, SNum + 1):          # rows
    for j in range(1, i + 1):      # numbers in each row
        print(j, end="")
    print()  # move to next line

# Cool Reverse Print Mechanism
for i in range(SNum, 0 , -1):          # rows
    for j in range(1, i + 1):      # numbers in each row
        print(j, end="")
    print()  # move to next line
```

**Functions & Recursion:**

    a) Write a recursive function to compute Fibonacci series up to n terms.
    b) Write another function to compute factorial using recursion and iteration.
    c) Compare both approaches in terms of efficiency.

```python
## a) Fibonacci using recursion (n terms) 
def fib(SNum):
    if SNum == 0:
        return 0
    elif SNum == 1:
        return 1
    else:
        return fib(SNum-1) + fib(SNum-2)

# print first n terms
in_num = int(input("Enter number of terms: "))

print("Fibonacci Series", in_num, ":")
for i in range(in_num):
    print(fib(i), end=" ")

# b) Factorial (recursion + iteration) : 

#    1. Recursive factorial -----------------------

def fact_recursive(SNum):
    if SNum == 0 or SNum == 1:
        return 1
    return SNum * fact_recursive(SNum - 1)

print("\nRecursive Factorial:", fact_recursive(in_num))

#     2. Iterative factorial  -----------------------

def fact_iterative(SNum):
    result = 1
    for i in range(2, SNum + 1):
        result *= i
    return result

print("\nIterative Factorial:", fact_iterative(in_num))
```

Recursive Fibonacci is inefficient due to repeated computations (O(2ⁿ)), while iterative methods are more efficient (O(n)). For factorial, both recursion and iteration take O(n) time, but iteration is preferred as it uses less memory.

Lists & Dictionaries: 
    a) Take a list of numbers and remove duplicates without using built-in set(). 
    b) Count frequency of each element using a dictionary. 
    c) Find the second largest element in the list.

```python
# data as input 

nums = [1, 2, 2, 3, 4, 4, 5]

print("Orignal list of Number: ", nums)

# a) Remove duplicates (without set())

uniqueA = []

for num in nums:
# put the number in unique numbers 
    if num not in uniqueA:
        uniqueA.append(num)

print("List without duplicates:", uniqueA)

# b) Count frequency using dictionary

freq = {}

for num in nums:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1

print("Frequency:", freq)

# c) Second largest element

uniqueB = []

for num in nums:
    if num not in uniqueB:
        uniqueB.append(num)

uniqueB.sort()

print("Second largest:", uniqueB[-2])
```

**5. Problem Solving (Mini Challenge):**

Given a list of student marks:
Marks = [78, 85, 62, 90, 55, 73, 88, 91, 67, 80]

a) Categorize students into grades using conditions.
b) Find top 3 students.
c) Display results in a structured format.
d) Extend logic to handle dynamic input size.

```python

# Static Marks DB Version of the program !  ---------------------

print("\nStatic Marks - Grading Functions\n")

# Static Marks DB 
Marks = [78, 85, 62, 90, 55, 73, 88, 91, 67, 80]
print("Static Marks DB:", Marks)

# Marks Grading Routine .
def get_grade(fm):
    if fm >= 90:
        return 'A'
    elif fm >= 75:
        return 'B'
    elif fm >= 60:
        return 'C'
    else:
        return 'D'

# Marks -> Grade Printing Logic
def marks_grade_print(fMarks):
    print("\nStudent Results:")
    print("----------------------")
    print("Student\tMarks\tGrade")

    for i in range(len(fMarks)):
       print(f"{i+1}\t{fMarks[i]}\t{get_grade(fMarks[i])}")

# Print the marks with grades as table 
marks_grade_print(Marks)

# Finding the Top 3 students
sorted_marks = sorted(Marks, reverse=True)

print("\nTop 3 Marks: Method 1")
for i in range(min(3, len(sorted_marks))):
    print(f"{i+1}: {sorted_marks[i]}")

# Static Marks DB END Version of the program !  ---------------------

# Dynamic Input version of the Program  ! --------------------------

print("\nDynamic Marks - Grading Functions\n")

# Dynamic input of the students and associate 
n = int(input("Enter number of students: "))

Marks = []
for i in range(n):
    m = int(input(f"Enter marks for student {i+1}: "))
    Marks.append(m)

print("Marks Database", Marks)

# Marks with Grade for Dynamic DB
marks_grade_print(Marks)

# Finding the Top 3 students
print("\nTop 3 Marks - Method 2",sorted(Marks, reverse=True)[:3])


```
