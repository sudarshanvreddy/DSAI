```python
## Q1 - Program
# Hotel Billing Program
# (c) - Sudarshan V | 25-APR-2026 
import random
from datetime import datetime 

"""
print(" Sudarshan Bhavan - Items ")
print("---------------------------")
print("SL- Item - Price")
print("1 - Idli - 10")
print("2 - Dosa - 25")
print("3 - Vada - 15")
print("4 - Coffee/Tea - 5")
print("--------------------------")
print(" All Prices in INR - Including GST")
"""

# Define the Items Database
# Add more if needed !

print("Welcome to IIIT Dharward - Canteen\n")


price_db = {
            1: {"item":"Idli", "price":10, "GST":5}, 
            2: {"item":"Dosa", "price":25, "GST":5},
            3: {"item":"Vada", "price":15, "GST":5},
            4: {"item":"Coffee/Tea", "price":5, "GST":3 },
}

# print(price_db)

# Display of the items in the database 

print("Number of Items in Menu:", len(price_db))

# MENU GUI ----------------------------------------- Start

print("-" * 33)
# print("| SL  | Item       | Price |") 
print(f"| {'Sl':<3} | {'Item':<15} | {'Price':<5} |")
print("-" * 33)

for item_nos in price_db:
#    print("\nItems:", price_db.items())
#    print("\nItem Nos:", item_nos)
#    print("|", item_nos, "  |", price_db[item_nos]["item"], "|", price_db[item_nos]["price"],"|")
    print(f"| {item_nos:<3} | {price_db[item_nos]["item"]:<15} | {price_db[item_nos]["price"]:<5} |")
print("-" * 33)

# MENU GUI ----------------------------------------- End

# Get the items that customer as choosen

print(" Items Consumed ")
cust_name = input("Customer Name")

# Init extreme / Max ID
item_ID = -1;

#Order DB
Cust_Ord = {}

#Order SL No
SlNo_Ord = 0;


# Start Taking the order 
while  item_ID != 0:
    #Get the Item Code
    items_ID = int(input("Item Code (0 for Exit)"))

    # Exit Loop by Entering - 0
    if items_ID == 0:
            break

    # Check if item is not in Dictionary 
    if items_ID not in price_db:
        print("Invalid ID Code")
        continue
        

    # Get How much Quantity Was Ordered
    item_Qty = int(input("Quantity Of Item"))

    Cust_Ord[items_ID] = item_Qty
# Order Done ----------------------

# Future Items
if len(Cust_Ord) != 0:
    print(Cust_Ord)
else:
    print("No Items in Order")


# GST Calculation

# Base Price = Final Price / (1 + (GST/100) ) 
# GST Amount = Final Price - Base Price 

Total_Amt = 0
Sl_No_Ord = 0
Total_Gst = 0

print("IIIT Dharward - Canteen\n")
print("Customer Name:", cust_name, "\nInvoice Number", random.randint(1, 1000), "\nDate/Time", datetime.now())
print("Customer Order")
print("-" * 94)
print(f"| {'Sl No'} | {'Item ID'} | {'Item Description':<8} | {'Unit Price'} | {'Quantity'} | {'Total'} | {'Base Amt/GST @GST Rate':<21} |")
print("-" * 94)


for IID, IQty in Cust_Ord.items():
    Sl_No_Ord += 1;

    #Total Price = Item Price x Qty 
    LineTotal = price_db[IID]["price"] * IQty
    #print("Line Item Price -", f"{LineTotal:.2f}")

    #Base Price = Line Item Price / (1 + (GST/100) ) 
    BaseTotal = LineTotal / (1 + price_db[IID]["GST"] / 100)
    #print("Line Item Base Price -", f"{BaseTotal:.2f}")

    # GST Amount = Final Price - Base Price 
    GSTTotal = LineTotal - BaseTotal
   # print("GST Amount -", f"{GSTTotal:.2f}")
    
    Total_Amt += LineTotal
    Total_Gst += GSTTotal

    print(f"| {Sl_No_Ord:<5} | {IID:<7} | {price_db[IID]["item"]:<16} | {price_db[IID]["price"]:<10} | {IQty:<8} | {LineTotal:<6}| {BaseTotal:.2f} + {GSTTotal:.2f} @ {price_db[IID]["GST"]}% {" ":<4} |")

print("-" * 94)

print("\nTotal Amount to be Paid : ", Total_Amt, "INR")
print("Total GST of the Bill : ", f"{Total_Gst:.2f}", "INR")

```


**Question 2**

Evaluate the quadratic equation for a given a, b, and c.

Quadratic Formula :

$x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}$

```python
import math

# Get the Variables

print("\nEnter the values a , b, c needed for Quadratic Equation")

while True:
    a = float(input("Enter a:"))
    
    if a == 0:
        print("a cannot be zero")
    else:
        break

        
b = float(input("Enter b:"))
c = float(input("Enter c:"))

#Calculation
print("\nThe Quadratic Equation to be solved", f"{a}x^2 + {b}x + {c} = 0", "| With Values : " "a =", a, "b =", b, " c=", c)

print("\nSolution to Quadratic Equation : ")

d = b**2 - 4*a*c # Calculate within the sq rt to check real v/s Img

if d > 0:
    x1 = (-b + math.sqrt(d)) / (2*a)
    x2 = (-b - math.sqrt(d)) / (2*a)
    print(f"Two real roots: {x1:.2f}, {x2:.2f}")

elif d == 0:
    x = -b / (2*a)
    print(f"One real root: {x:.2f}")

else:
    real = -b / (2*a)
    imag = math.sqrt(-d) / (2*a)
    print(f"Complex roots: {real:.2f} + {imag:.2f}i and {real:.2f} - {imag:.2f}i")

```


**Question 4**

Write a Python program that computes the factorial of an integer

## Factorial Definition

The factorial of a non-negative integer \( n \) is defined as:

$$
n! = n \times (n-1) \times (n-2) \times \dots \times 1
$$

For example:

$$
5! = 5 \times 4 \times 3 \times 2 \times 1 = 120
$$

By definition:

$$
0! = 1
$$

```python
# Program for factorial Calculation

print("Program for factorial Calculation")

hold_num = 1

while True:
    fac_base = int(input("Enter an non-negative integer:"))

    if fac_base < 0:
        print("Enter a non-negative integer:")
        continue
    elif fac_base == 0:
        print(f"Factorial of {fac_base} is 1")
        break
    else:
        for i in range(fac_base, 1, -1):
            hold_num *= i
        break

print(f"The Factorial of {fac_base} is {hold_num} ")
                   
                   
```

**Question 8 , 9 and 10**

8.	Create a dataframe with name, age and marks.
9.	Filter students with marks > 80, sort by age.
10.	Export the filtered  Dataframe to csv.


```python
# Program for Question 8 , 9 and 10
import numpy as np
import pandas as pd

# Step 1 , i.e Question 8 : Create a dataframe with Name, Age and Marks
myData = {
    "SName" : ["Sudarshan", "Ramesh", "Mayank", "Sunil"],
    "SAge" : [41, 35, 26, 22],
    "SMarks" : [99, 65, 83, 85]
}

print("Original Data\n",myData)

my_df = pd.DataFrame(myData)

print("\nOriginal DataFrame\n",my_df)


# Step 2 , i.e Question 9 : Filter students with marks > 80, sort by age.

fil_df = my_df[my_df["SMarks"] > 80].sort_values(by="SAge")

print("\nFiltered and Sorted DataFrame:")
print(fil_df)


# Step 3,  i.e Question 10,  Export the filtered DataFrame to CSV

print("\nStoring Filtered data -> f_stu_1.csv")
fil_df.to_csv("f_stu_1.csv", index=False)


```
